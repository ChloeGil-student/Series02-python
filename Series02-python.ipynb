# Series02-Python : Sinusoidal signals
The goal of this notebook is to plot sinusoidal signal as part of the M1 IEAP program at the University of Montpellier. 

x(t) = Asin(ω·t + φ)

A = amplitude (unit of the physical system we simulate)
f = frequency 
w = angular frequency (rad/s) = 2*pi*f 
phi = phase when t = 0 (rad) 

```python
# Import the needed libraries
import numpy as np 
import matplotlib.pyplot as plt 

# Create a numpy array called t that contains 1000 values from 0 to 1 (t is for time)
t = np.linspace(0,1,1000)

# Create a numpy array called s that contains the values of a sine function of t (s is for the signal)
s = np.sin(2*np.pi*2*t)

# Plot s as a function of t
plt.plot(t,s)

# Add a title to the plot
plt.title('Sine wave')

# Add label to the x-axis 
plt.xlabel('Time (s)')

# Add label to the y-axis
plt.ylabel('Amplitude')

# Add a grid to the plot
plt.grid(True)

# Save the plot as a PNG file
plt.savefig('Sinusoidal function.png')
```

```python
# Create a numpy array called t that contains 1000 values from 0 to 1 (t is for time)
t = np.linspace(0,1,1000)

# Create a numpy array called s that contains the values of a sine function of t (s is for the signal)
s = np.sin(2*np.pi*t) 

# Plot s as a function of t
plt.plot(t,s, label='Sinus')

# Add a title to the plot
plt.title('Sinusoidal function')

# Add label to the x-axis 
plt.xlabel('Time (s)')

# Add label to the y-axis
plt.ylabel('Amplitude')

# Add a grid to the plot
plt.grid(True)

# Add a second plot of a cosine function of t in the same figure
plt.plot(t, np.cos(2*np.pi*t), label='Cosinus')

# Add a legend to the plot
plt.legend()

# Save the plot as a pdf file
plt.savefig('Sinusoidal function.pdf')
```

```python
def sinusoid (A, omega, phi, t):
    """
    Function to generate a sinusoidal signal. x(t) = A * sin(omega * t + phi)
    
    Parameters:
    A : float
        Amplitude of the sinusoid.
    omega : float
        Angular frequency of the sinusoid in radians/second (omega = 2*pi*f).
    phi : float
        Phase shift of the sinusoid, in radians.
    t : array-like
        Time values at which to evaluate the sinusoid.
        
    Returns:
    x : array-like
        Sinusoidal signal evaluated at time t.
    """
    x = A * np.sin(omega * t + phi) 
    return x

# Time vector from 0 to 1 second, with 1000 samples
t = np.linspace(0, 1, 1000)

# Two different frequencies for the sinusoidal signals
f1 = 1  # Frequency of the first sinusoid in Hz
f2 = 2  # Frequency of the second sinusoid in Hz

# Convert frequencies to angular frequencies (in radians/second) (since sinusoid expects rad/s not f)
omega1 = 2 * np.pi * f1
omega2 = 2 * np.pi * f2

# Generate the sinusoidal signals
s1 = sinusoid(1, omega1, 0, t)
s2 = sinusoid(1, omega2, 0, t)

# Plot 
plt.plot(t,s1, label='Sinus 1')
plt.plot(t,s2, label='Sinus 2')
plt.title('Sinusoidal function')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)
plt.show()
```
